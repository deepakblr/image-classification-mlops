# Cats vs Dogs — Exploratory Notebook

Quick look at class balance and a few sample images after running `scripts/download_data.py`.

```python
from pathlib import Path
from collections import Counter
from PIL import Image
import matplotlib.pyplot as plt

from src.data.preprocess import list_class_images, load_splits

images = list_class_images()
print({k: len(v) for k, v in images.items()})

fig, axes = plt.subplots(2, 4, figsize=(10, 5))
for row, cls in enumerate(["cat", "dog"]):
    for col, path in enumerate(images[cls][:4]):
        axes[row, col].imshow(Image.open(path))
        axes[row, col].set_title(cls)
        axes[row, col].axis("off")
plt.tight_layout()
plt.show()
```
